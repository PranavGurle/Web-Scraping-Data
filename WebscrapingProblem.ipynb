```python
import requests
```

```python
URL = "https://quotes.toscrape.com/page/1/"

res = requests.get(URL)
```

```python
print(res.status_code)
```

```python
if res.status_code == 200:
    print(res.headers)
```

```python
with open("scraped_data/data2.html", "w", encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup


with open("scraped_data/data2.html", "r", encoding="utf-8") as f:
    html_content= f.read()


soup = BeautifulSoup(html_content, "lxml")
    
```
